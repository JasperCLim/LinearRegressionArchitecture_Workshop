```python
from datetime import datetime, timezone
import logging
from pathlib import Path
from typing import Any

import pandas as pd

from src.data_loader import load_csv, load_from_api, load_from_sqlite


class DataExtractionAnalysis:
    def __init__(self, config: dict):
        self.config = config
        self.source_type = config.get("source_type", "csv").lower()
        self.env = config.get("env", "dev")
        self.logger = self._setup_logger()
        self.profile = {}

    def _setup_logger(self) -> logging.Logger:
        log_path = Path(self.config.get("log_path", "experiments/data_extraction.log"))
        log_path.parent.mkdir(parents=True, exist_ok=True)
        logger = logging.getLogger("DataExtractionAnalysis")
        logger.setLevel(logging.INFO)
        logger.propagate = False
        if not logger.handlers:
            handler = logging.FileHandler(log_path, encoding="utf-8")
            handler.setFormatter(logging.Formatter("%(asctime)s - %(levelname)s - %(message)s"))
            logger.addHandler(handler)
        return logger

    def detect_source(self) -> None:
        supported_sources = {"csv", "api", "sqlite"}
        if self.source_type not in supported_sources:
            raise ValueError(f"Unsupported source type: {self.source_type}")
        self.logger.info("Using %s source in %s environment", self.source_type, self.env)

    def connect_to_source(self) -> pd.DataFrame:
        if self.source_type == "csv":
            return load_csv(self.config["path"])
        if self.source_type == "api":
            return load_from_api(
                self.config["url"],
                params=self.config.get("params"),
                headers=self.config.get("headers"),
            )
        return load_from_sqlite(
            self.config["db_path"],
            self.config["query"],
            params=self.config.get("params"),
        )

    def validate_schema(self, data: Any) -> bool:
        if not isinstance(data, pd.DataFrame):
            return False
        required_columns = self.config.get("required_columns", [])
        missing_columns = set(required_columns) - set(data.columns)
        if missing_columns:
            self.logger.error("Missing columns: %s", sorted(missing_columns))
            return False
        return True

    def snapshot_data(self, data: pd.DataFrame) -> None:
        snapshot_path = Path(
            self.config.get("snapshot_path", "experiments/source_snapshot.csv")
        )
        snapshot_path.parent.mkdir(parents=True, exist_ok=True)
        data.to_csv(snapshot_path, index=False)
        self.logger.info("Saved %s rows to %s", len(data), snapshot_path)

    def profile_data(self, data: pd.DataFrame) -> dict:
        profile = {
            "profiled_at_utc": datetime.now(timezone.utc).isoformat(),
            "rows": len(data),
            "columns": list(data.columns),
            "missing_values": data.isna().sum().to_dict(),
            "dtypes": data.dtypes.astype(str).to_dict(),
        }
        self.logger.info("Profiled %s rows and %s columns", len(data), len(data.columns))
        return profile

    def extract(self) -> pd.DataFrame:
        self.detect_source()
        data = self.connect_to_source()
        if not self.validate_schema(data):
            raise ValueError("Input data failed schema validation")
        self.snapshot_data(data)
        self.profile = self.profile_data(data)
        return data
```
